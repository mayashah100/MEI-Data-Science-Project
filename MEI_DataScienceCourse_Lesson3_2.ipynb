## Plotting scatter diagrams

You are probably used to drawing scatter diagrams to look for association between two variables. You can use Seaborn's `relplot()` (relational plot) command to create a scatter diagram. The format of the `relplot()` command should look very similar to `catplot()` and `displot()`.

> Run the code below to draw a scatter diagram of `Life expectancy at birth 2010` against `GDP per capita (US$)`.

```python
# create a scatter plot of life expectancy against GDP
sns.relplot(data=countries_data, x='GDP per capita (US$)', y='Life expectancy at birth 2010', aspect=2);
```

> Add and run code below to:
> * Create a scatter plot of life expectancy against unemployment;
> * Create a scatter plot of life expectancy against birth rate.

```python
import matplotlib.pyplot as plt

# create a scatter plot of life expectancy against unemployment
g = sns.relplot(data=countries_data, x='unemployment (%)', y='Life expectancy at birth 2010', aspect=2)

# Find the country with the highest unemployment rate for annotation
highest_unemployment_country = countries_data.loc[countries_data['unemployment (%)'].idxmax()]
country_name = highest_unemployment_country['Country']
unemployment_val = highest_unemployment_country['unemployment (%)']
life_expectancy_val = highest_unemployment_country['Life expectancy at birth 2010']

# Annotate the specific point on the plot
# g.axes is a 2D array of axes, g.axes[0,0] gets the first (and only) subplot's axes
g.axes[0, 0].annotate(country_name,
                     xy=(unemployment_val, life_expectancy_val),
                     xytext=(unemployment_val + 2, life_expectancy_val), # Offset text slightly for readability
                     arrowprops=dict(facecolor='black', shrink=0.05),
                     ha='left', va='center')

g.axes[0, 0].set_title('Life Expectancy vs. Unemployment Rate with Highest Unemployment Country Highlighted')
g.axes[0, 0].grid(True, linestyle='--', alpha=0.7)
```

```python
# create a scatter plot of life expectancy against birth rate
sns.relplot(data=countries_data, x='birth rate per 1000', y='Life expectancy at birth 2010', aspect=2)
```

To filter the data, let's first determine suitable thresholds for 'long life expectancy' and 'low GDP'. We'll use the 75th percentile for life expectancy (to define 'long') and the 25th percentile for GDP (to define 'low').

```python
life_expectancy_75th_percentile = countries_data['Life expectancy at birth 2010'].quantile(0.75)
gdp_25th_percentile = countries_data['GDP per capita (US$)'].quantile(0.25)

print(f"75th percentile for Life Expectancy: {life_expectancy_75th_percentile:.2f}")
print(f"25th percentile for GDP per capita: {gdp_25th_percentile:.2f}")

# Now, filter the data
long_life_low_gdp_countries = countries_data[
    (countries_data['Life expectancy at birth 2010'] > life_expectancy_75th_percentile) &
    (countries_data['GDP per capita (US$)'] < gdp_25th_percentile)
]

print("\nCountries with long life expectancy and relatively low GDP:")
display(long_life_low_gdp_countries[['Country', 'Life expectancy at birth 2010', 'GDP per capita (US$)']])
```

## Hexagonal bin plots

When you have a very large data set, with thousands or even millions of data points, scatter plots can be misleading. This is because the points start landing on top of each other, making it hard to tell if there are many points in an area or just a few. To solve this, some charts 'bin' the data first, in much the same way a histogram does. Unlike a histogram, a bin plot shows density using shading with colours representing higher densities.

Seaborn can create a bin plot using its `jointplot` function, which also provides histograms of the two variables.

> Run the code below to create a hexagonal bin plot of life expectancy against GDP

```python
# create a hexagonal bin plot
sns.jointplot(data=countries_data, kind='hex', x='GDP per capita (US$)', y='Life expectancy at birth 2010');
```

Let's re-plot the hexagonal bin plot with a logarithmic scale for 'labor force' to better visualize the distribution, as the 'labor force' variable typically has a very wide range and is often right-skewed. We'll also set a `height` for better visualization.

Given the discussion about 'labor force', let's explore another demographic rate: **'death rate per 1000'** against 'Life expectancy at birth 2010' using a hexagonal bin plot. This should provide insights into how a lower death rate correlates with higher life expectancy.

```python
# Create a new column with the log-transformed 'labor force' to handle its skewed distribution
countries_data['labor force (log)'] = countries_data['labor force'].apply(lambda x: np.log10(x) if x > 0 else 0)

# Create a hexagonal bin plot with log-transformed 'labor force'
# Use the 'height' parameter to control the overall size of the plot
sns.jointplot(data=countries_data, kind='hex', x='labor force (log)', y='Life expectancy at birth 2010', height=8)

# Note: If there are zero or negative 'labor force' values, the log transformation will produce errors.
# The lambda function handles values <= 0 by setting them to 0 (or a small number like 1 which log10 is 0).
# You might consider imputation or dropping such rows if they are present and problematic for your analysis.
```

```python
# Create a hexagonal bin plot of life expectancy against death rate per 1000
sns.jointplot(data=countries_data, kind='hex', x='death rate per 1000', y='Life expectancy at birth 2010', height=8)
plt.suptitle('Life Expectancy vs. Death Rate per 1000 Population (Hexagonal Bin Plot)', y=1.02) # Add a title to the figure
plt.show()
```

### Analyzing the impact of Land Borders on Life Expectancy

Let's visualize the distribution of life expectancy for countries with and without land borders using a box plot.
